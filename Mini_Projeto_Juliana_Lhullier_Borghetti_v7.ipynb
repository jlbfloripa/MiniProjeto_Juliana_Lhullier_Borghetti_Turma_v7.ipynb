```python
# Usei a lógica (if/else) para preencher as categorias com "#N/D" substituindo por "Sem Categoria""

categorias_corrigidas = []

for categoria in df_varejo_limpo["categoria_produto"]:
    if categoria == "#N/D":
        categorias_corrigidas.append("Sem categoria")
    else:
        categorias_corrigidas.append(categoria)

df_varejo_limpo["categoria_produto"] = categorias_corrigidas
df_varejo_limpo["categoria_produto"].value_counts(dropna=False)
```

```python
# Usei a lógica (if/else) para preencher as categorias com "#N/D" substituindo por "Sem Categoria" na coluna "nome_produto"

categorias_corrigidas_produto = []

for categoria in df_varejo_limpo["nome_produto"]:
    if categoria == "#N/D":
        categorias_corrigidas_produto.append("Sem categoria")
    else:
        categorias_corrigidas_produto.append(categoria)

df_varejo_limpo["nome_produto"] = categorias_corrigidas_produto
df_varejo_limpo["nome_produto"].value_counts(dropna=False)
```
