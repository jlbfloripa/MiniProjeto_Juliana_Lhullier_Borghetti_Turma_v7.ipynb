 ## Estatísticas descritivas básicas 

```python
# Pedi o número de linhas da coluna "numero_filhos" para iniciar a análise estatística descritiva básica proposta para o mini-projeto

len(df_varejo_limpo["numero_filhos"])
```

```python
# Verifiquei a quantidade de clientes únicos 

df_varejo_limpo["ID_cliente"].nunique()
```

```python
# Agrupei um único número de filhos por cliente usando o groupby() e o first() 

df_filhos = df_varejo_limpo.groupby("ID_cliente")["numero_filhos"].first()
```

```python
# Contei os clientes por número de filhos usando value_counts() ordenando com sort_index()

df_filhos.value_counts().sort_index()
```

```python
# Usei o describe() para gerar estatísticas descritivas (média; mediana; desvio padrão; máximo; mínimo; e contagem, quartis)

df_filhos.describe()
```

```python
# Usei mode() para obter a moda

df_filhos.mode()
```

```python
# Para melhor visualização, fiz a mesma operação com print()

print("Estatísticas do número de filhos por cliente:")
print("Contagem:", df_filhos.count())
print("Média:", df_filhos.mean().round(2))
print("Mediana:", df_filhos.median().round(2))
print("Moda:", df_filhos.mode().tolist())
print("Desvio padrão:", df_filhos.std().round(2))
print("Mínimo:", df_filhos.min())
print("Máximo:", df_filhos.max())
print("Quartis:")
print(df_filhos.quantile([0.25, 0.50, 0.75]))
```

```python
# Calculei a proporção de clientes de cada grupo que comprou cada categoria

'''
Criei uma tabela com pivot_table() cruzando duas variáveis - número de filhos e categoria do produto;
Usei index para colocar o número de filhos nas linhas;
Coloquei a categoria do produto na coluna com columns;
Usei o aggfunc= "size" para contar os registros em cada combinação filhos-categoria;
Pedi para colocar o número zero quando não houvesse registro com fill_value = 0. 


'''
tabela = df_varejo_limpo.pivot_table(
    index = "numero_filhos",
    columns = "categoria_produto",
    aggfunc = "size",
    fill_value = 0
)

preferencia = tabela.div(tabela.sum(axis=1), axis=0) * 100

print(preferencia.round(2))
```
