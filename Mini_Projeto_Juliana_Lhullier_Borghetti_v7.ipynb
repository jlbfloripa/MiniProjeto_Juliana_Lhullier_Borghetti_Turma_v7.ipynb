# Limpeza de dados

```python
# Criei uma cópia do csv

df_varejo_limpo = df_varejo.copy()
```

```python
# Renommei as colunas

df_varejo_limpo = df_varejo_limpo.rename(columns={
    "DATA": "data",
    "CO_ID": "ID_compra",
    "CL_ID": "ID_cliente",
    "CL_GENERO": "genero_cliente",
    "CL_EC": "estado_civil",
    "CL_FHL": "numero_filhos",
    "CL_SEG": "classe_economica",
    "PR_ID": "codigo_produto",
    "PR_CAT": "categoria_produto",
    "PR_NOME": "nome_produto"
})

# Verifiquei

df_varejo_limpo.head()
```

```python
# Verifiquei então se todas as colunas estavam corretas

print(df_varejo_limpo.columns.tolist())
```

```python
# Verifiquei os valores ausentes

df_varejo_limpo.isna()
```

```python
# Verifiquei todas (all) as colunas vazias (isna) e criei uma variável (colunas_vazia) para armazenar esses resultados

colunas_vazias = df_varejo_limpo.columns[df_varejo_limpo.isna().all()]

# Imprimi a contagem (len) e o nome das colunas vazias

print("Total de colunas vazias: ", len(colunas_vazias))
print("\nNomes das colunas vazias: ")

#Por fim, pedi para listar essas colunas

print(colunas_vazias.tolist())
```

```python
# Removi as colunas totalmente vazias

df_varejo_limpo = df_varejo_limpo.dropna(axis=1, how="all")

#Verifiquei as colunas restantes
print(df_varejo_limpo.columns.tolist())
```

```python
# Criei variáveis para armazenar os valores duplicados ou únicos
# Separei as linhas duplicadas para análise

mascara_duplicadas = df_varejo_limpo.duplicated(keep=False)

linhas_duplicadas = df_varejo_limpo[mascara_duplicadas].copy()

print("Linhas duplicadas: ", len(linhas_duplicadas))
print("Linhas únicas: ", len(df_varejo_limpo.drop_duplicates()))
```
